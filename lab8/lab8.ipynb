```
from pyspark.sql.functions import col, explode
```

```
json_path = "dbfs:/FileStore/tables/brzydki.json"

df = spark.read.option("multiline", "true").json(json_path)

features_df = df.select(explode("features").alias("feature"))

# wyciąganie pól z 'properties'
flattened_df = features_df.select(
    col("feature.properties.featureID").alias("featureID"),
    col("feature.properties.toid").alias("toid"),
    col("feature.properties.alternativeToids").alias("alternativeToids"),
    col("feature.properties.changeEventType").alias("changeEventType"),
    col("feature.properties.jobReference").alias("jobReference"),
    col("feature.properties.validFromTimestamp").alias("validFromTimestamp"),
    col("feature.properties.changeEventType").alias("machangeEventTyperried"),
    col("feature.properties.jobReference").alias("jobReference"),
    col("feature.properties.lifecycleStatusComponent").alias("lifecycleStatusComponent"),
    col("feature.properties.linearHeightComponent").alias("linearHeightComponent"),
)

flattened_df.show(truncate=False)
```
