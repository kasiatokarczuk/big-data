```
from pyspark.sql.functions import col, rand

# przykładowy df
df = spark.range(0, 10000) \
    .withColumn("category", (rand() * 5).cast("int")) \
    .withColumn("value", (rand() * 100).cast("int"))

df.write.mode("overwrite").saveAsTable("analyze_demo_table")
```

```
%sql
-- Statystyki dla całej tabeli
ANALYZE TABLE analyze_demo_table COMPUTE STATISTICS;

-- Statystyki dla wszystkich kolumn
ANALYZE TABLE analyze_demo_table COMPUTE STATISTICS FOR COLUMNS category, value;
```

```
spark.sql("DESCRIBE TABLE EXTENDED analyze_demo_table").show(truncate=False)
spark.catalog.listColumns("analyze_demo_table")
```
