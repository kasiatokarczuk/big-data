```
from pyspark.sql.functions import col

# Wczytanie danych JSON
df = spark.read.json("dbfs:/FileStore/tables/Nested.json",multiLine=True)

# sprawdzenie struktury danych
df.printSchema()
```

```
# Usuwanie 'attribute1' z poziomu głównego
df = df.drop("attribute1")

# Usuwanie pola z wewnętrznych struktur
df = df.withColumn(
    "pathLinkInfo",
    col("pathLinkInfo").dropFields("endGradeSeparation", "formsPartOfPath")
).withColumn(
    "pathLinkInfo",
    col("pathLinkInfo").withField(
        "elevationGain",
        col("pathLinkInfo.elevationGain").dropFields("elevationAgainstDirection")
    )
)

# Podgląd danych
df.printSchema()
df.show(truncate=False)
```
